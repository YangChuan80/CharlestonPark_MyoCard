```python
import tkinter as tk
from tkinter import ttk

root = tk.Tk()
root.title("选项卡示例")
root.geometry("400x300")
```

##### 创建 Notebook（选项卡容器）

```python
notebook = ttk.Notebook(root)
notebook.pack(fill='both', expand=True)
```

##### 创建各个选项卡的 Frame

```python
tab1 = ttk.Frame(notebook)
tab2 = ttk.Frame(notebook)
tab3 = ttk.Frame(notebook)
```

##### 把 Frame 添加到 Notebook，并指定标签文字

```python
notebook.add(tab1, text="选项卡 1")
notebook.add(tab2, text="选项卡 2")
notebook.add(tab3, text="选项卡 3")
```

##### 在选项卡中放置控件

```python
ttk.Label(tab1, text="这是第一个选项卡").pack(pady=20)
```

```python
ttk.Button(tab2, text="按钮").pack(pady=20)
```

```python
ttk.Entry(tab3).pack(pady=20)
```

```python
root.mainloop()
```
